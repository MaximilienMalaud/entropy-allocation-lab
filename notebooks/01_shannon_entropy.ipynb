# Phase 1, lesson 1: Shannon entropy

$$H = -\sum_i p_i \log_2(p_i)$$

Entropy is the **average surprise**, weighted by the probability of each outcome. It ranges from 0 (certainty) to $\log_2(n)$ (all $n$ outcomes equally likely). It ignores the **order** of observations: shuffling a series does not change its entropy.

## Step 2: the function

```python
def shannon_entropy(probabilities):
    """Shannon entropy, in bits, of a probability distribution."""
    p = np.asarray(probabilities, dtype=float)
    p = p[p > 0]  # an impossible outcome adds no average surprise
    return -np.sum(p * np.log2(p))
```

## Step 4: mini-challenge, normalised entropy

Write `normalized_shannon_entropy(probabilities)`, which returns $H / \log_2(n)$.

Watch out: $n$ is the **total** number of possible outcomes, zeros included (`[1, 0, 0]` has $n = 3$).

Expected: `[0.2, 0.3, 0.5]` gives 0.937.

```python
def normalized_shannon_entropy(probabilities):
    # your code here
    n = len(probabilities)
    h = shannon_entropy(probabilities)
    return h/np.log2(n)

print(normalized_shannon_entropy([0.2, 0.3, 0.5]))
print(normalized_shannon_entropy([0.5, 0.5]))
print(normalized_shannon_entropy([1, 0, 0]))
```
